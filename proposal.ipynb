## Dataset description
* Source: https://huggingface.co/datasets/openfoodfacts/product-database

* The dataset is in parqeut format, and it contains millions of food products, including nutritional information, ingredients, product names, and food categories

* Description: The dataset will contain approximately 1,000 candy products from Open Food Facts, with at least 10 attributes describing their nutritional information and ingredients. These attributes will include numerical variables such as calories, total fat, saturated fat, carbohydrates, sugar, fiber, protein, and salt content per 100 grams. It will also include categorical (binary) variables indicating whether a candy contains specific ingredients, such as cocoa, milk, gelatin, pectin, citric acid, and peanuts. The class attribute will be "candy_type", which will classify each product into one of four categories: Gummy, Chocolate, Caramel, or Chewing Gum. The goal of the project will be to predict a candy's type based on its nutritional characteristics and ingredients.


## Implementation/technical merit
* Anticipated challenges: Very large dataset (5 Million rows) because it tracks data on all foods. So we need to filter data for just candy to begin with. Afterwards we will need to clean the data and review we haven't lost anything during the candy isolation process. We anticipate it will be difficult to clean it, we plan to go from 5 million rows to 1000. To pare down the attributes we will be using set standards for what qualifies an instance to be candy, as well as unessesary attributes. 


## Potential impact of the results
* These results are useful because it would show that a machine learning model can classify a food based on its attributes rather than on a physical picture of the food. Groups most impacted by this would be the candy manufacturers, because they can compare their version of a certain classification with competitors to determine which consumers like more.
